# 02 — VH Backscatter Threshold Baseline

Tunes a single Sentinel-1 VH dB threshold strictly on the `valid` split, then evaluates IoU, F1, Precision, and Recall on `test` and held-out `bolivia`.

```python
import json
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve().parent))

from ml.src.dataset import download_official_splits, load_split_pairs
from ml.src.evaluate import eval_vh_baseline, tune_vh_threshold_on_val

splits = download_official_splits()
val_pairs = load_split_pairs(splits['valid'])
test_pairs = load_split_pairs(splits['test'])
bolivia_pairs = load_split_pairs(splits['bolivia'])

best_vh_db = tune_vh_threshold_on_val(val_pairs)
print(f'Tuned VH threshold on validation split: {best_vh_db} dB')

for split_name, pairs in [('valid', val_pairs), ('test', test_pairs), ('bolivia', bolivia_pairs)]:
    metrics = eval_vh_baseline(pairs, best_vh_db)
    print(split_name, json.dumps(metrics, indent=2))
```
