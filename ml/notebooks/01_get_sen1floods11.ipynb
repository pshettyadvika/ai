```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve().parent))

from ml.config import SEN1FLOODS11_DIR
from ml.src.dataset import download_official_splits, load_split_pairs

(SEN1FLOODS11_DIR / 'S1Hand').mkdir(parents=True, exist_ok=True)
(SEN1FLOODS11_DIR / 'LabelHand').mkdir(parents=True, exist_ok=True)
splits = download_official_splits()
print('Downloaded official split CSVs:', splits)
```
